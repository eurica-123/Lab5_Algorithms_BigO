```python
# STUDENT-SPECIFIC CONFIGURATION & INPUT RULES

LAST_NAME = "dela Cruz"
STUDENT_ID = "TUPM-26-1839"
FAVORITE_ARTIST = "Shirebound and Busking"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.upper()) - ord('A') + 1 for c in FAVORITE_ARTIST if c.isalpha())

print("--- CONFIGURATION LOADED ---")
print(f"Seed Digit: {SEED_DIGIT}")
print(f"ID Sum: {ID_SUM}")
print(f"Name Length: {NAME_LENGTH}")
print(f"Artist Length: {ARTIST_LENGTH}")
print(f"Artist Sum: {ARTIST_SUM}\n")
```

```python
# EXERCISE 1: Emergency Dispatch Record Search

def search_rotated(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        if data[low] <= data[mid]:
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

def linear_search_rotated(data, target):
    for i, val in enumerate(data):
        if val == target:
            return i, i + 1
    return -1, len(data)

original_dataset = sorted(list(set([(ID_SUM * 3 + i * 5) % 1000 for i in range(20)])))
rotation_index = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_index:] + original_dataset[:rotation_index]
target_incident = (ID_SUM + ARTIST_SUM) % 1000

index, comp_rotated = search_rotated(rotated_dataset, target_incident)
lin_index, comp_linear = linear_search_rotated(rotated_dataset, target_incident)

print("=== EXERCISE 1: ROTATED RECORD SEARCH ===")
print("Rotated Dataset:", rotated_dataset)
print("Target Incident:", target_incident)
print(f"Rotated Search -> Index: {index}, Comparisons: {comp_rotated}")
print(f"Linear Search  -> Index: {lin_index}, Comparisons: {comp_linear}\n")
```

```python
# EXERCISE 2: University Enrollment Records

dept_a = sorted([(SEED_DIGIT * 7 + i * 4) % 200 for i in range(8)])
dept_b = sorted([(ARTIST_LENGTH * 5 + i * 6) % 200 for i in range(8)])

def merge_sorted_lists(list1, list2):
    i, j, comparisons = 0, 0, 0
    merged = []
    while i < len(list1) and j < len(list2):
        comparisons += 1
        if list1[i] <= list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1
    merged.extend(list1[i:])
    merged.extend(list2[j:])
    return merged, comparisons

consolidated, merge_comparisons = merge_sorted_lists(dept_a, dept_b)

print("=== EXERCISE 2: ENROLLMENT CONSOLIDATION ===")
print("Department A:", dept_a)
print("Department B:", dept_b)
print("Consolidated:", consolidated)
print(f"Records Before: {len(dept_a) + len(dept_b)} | Records After: {len(consolidated)}")
print(f"Comparison Count: {merge_comparisons}")
print(f"Ordering Verified: {consolidated == sorted(consolidated)}\n")
```

```python
# EXERCISE 3: E-Commerce Order Processing

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

raw_orders = [(ID_SUM * 2 + i * 13) % 500 for i in range(20)]
conditions = {
    "Sorted": sorted(raw_orders),
    "Reverse-Ordered": sorted(raw_orders, reverse=True),
    "Random": raw_orders.copy()
}

print("=== EXERCISE 3: E-COMMERCE SORTING COMPARISON ===")
print(f"{'Condition':<16} | {'Algorithm':<16} | {'Comparisons':<12} | {'Movements':<10}")
print("-" * 62)
for cond_name, dataset in conditions.items():
    _, b_comp, b_swap = bubble_sort(dataset)
    _, s_comp, s_swap = selection_sort(dataset)
    _, i_comp, i_shift = insertion_sort(dataset)
    
    print(f"{cond_name:<16} | {'Bubble Sort':<16} | {b_comp:<12} | {b_swap:<10}")
    print(f"{'':<16} | {'Selection Sort':<16} | {s_comp:<12} | {s_swap:<10}")
    print(f"{'':<16} | {'Insertion Sort':<16} | {i_comp:<12} | {i_shift:<10}")
    print("-" * 62)
```
